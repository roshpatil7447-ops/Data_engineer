```
%%sql
use database snowflake_sample_data;
use schema tpch_sf1;
```

```
from snowflake.snowpark.context import get_active_session

session = get_active_session()
df=session.table("CUSTOMER")
```

```
df.show(10)
```

```
df1 = df.select("C_MKTSEGMENT").distinct()
df1.show()
```

```
from snowflake.snowpark.functions import col

df2=df.select("C_CUSTKEY","C_NAME","C_NATIONKEY","C_PHONE","C_ACCTBAL","C_MKTSEGMENT")
df3=df2.filter(col("C_MKTSEGMENT")== "AUTOMOBILE")
df3.show()
```

```
df3.write.mode("overwrite").save_as_table("SAMPLE_DB.SAMPLE_SCH.customer_data")
```
